```python
def print_processing_statistics(stats_file_path=None):
    """
    Load and print processing statistics from JSON file.

    Parameters:
    -----------
    stats_file_path : str or Path, optional
        Path to the statistics JSON file. If None, uses config.OUTPUT_FOLDER/processing_statistics.json
    """
    import json
    from pathlib import Path
    from datetime import datetime

    # Determine stats file path
    if stats_file_path is None:
        stats_file = Path(config.OUTPUT_FOLDER) / "processing_statistics.json"
    else:
        stats_file = Path(stats_file_path)

    if not stats_file.exists():
        print(f"❌ Statistics file not found: {stats_file}")
        print("Run main.py first to generate the statistics file.")
        return None

    # Load statistics
    with open(stats_file, "r") as f:
        stats = json.load(f)

    # Helper function to safely get nested dict values
    def get_nested(data, *keys, default="N/A"):
        """Safely get nested dictionary values."""
        try:
            for key in keys:
                data = data[key]
            return data
        except (KeyError, TypeError):
            return default

    # Helper to convert unix timestamp to UTC
    def unix_to_utc(timestamp):
        """Convert unix timestamp to UTC string."""
        try:
            return datetime.utcfromtimestamp(float(timestamp)).strftime(
                "%Y-%m-%d %H:%M:%S UTC"
            )
        except:
            return "N/A"

    print("=" * 80)
    print("PROCESSING STATISTICS")
    print("=" * 80)

    # Session info
    print("\n📅 SESSION INFO")
    print(f"  Started: {get_nested(stats, 'processing_session', 'start_time')}")
    print(f"  Finished: {get_nested(stats, 'processing_session', 'end_time')}")
    duration_sec = get_nested(stats, "processing_session", "duration_sec", default=0)
    duration_human = get_nested(
        stats, "processing_session", "duration_human", default="N/A"
    )
    if duration_sec != "N/A" and duration_sec != 0:
        print(f"  Processing duration: {duration_human}")

    # Configuration
    print("\n⚙️  CONFIGURATION")
    print(
        f"  Navigation CSV: {Path(get_nested(stats, 'configuration', 'nav_csv', default='')).name}"
    )
    print(
        f"  MBES GeoTIFF: {Path(get_nested(stats, 'configuration', 'mbes_geotiff', default='')).name}"
    )
    print(
        f"  Output folder: {Path(get_nested(stats, 'configuration', 'output_folder', default='')).name}"
    )
    print(
        f"  Max ray length: {get_nested(stats, 'configuration', 'max_ray_length_m', default=0):.1f} m"
    )
    print(
        f"  Failure threshold: {get_nested(stats, 'configuration', 'early_failure_threshold_pct', default=0):.1f}%"
    )

    # Navigation data
    print("\n🧭 NAVIGATION DATA")
    nav_records = get_nested(stats, "navigation_data", "total_records", default=0)
    print(f"  Total records: {nav_records:,} (navigation data points in CSV)")
    time_start = get_nested(stats, "navigation_data", "time_range", "start", default=0)
    time_end = get_nested(stats, "navigation_data", "time_range", "end", default=0)
    time_dur = get_nested(
        stats, "navigation_data", "time_range", "duration_sec", default=0
    )
    print(f"  Start time: {time_start:.2f} ({unix_to_utc(time_start)})")
    print(f"  End time: {time_end:.2f} ({unix_to_utc(time_end)})")
    print(f"  Duration: {time_dur:.1f} s ({time_dur/60:.1f} min)")

    # MBES mesh
    print("\n🗺️  MBES MESH")
    print(f"  Vertices: {get_nested(stats, 'mbes_mesh', 'n_vertices', default=0):,}")
    print(f"  Faces: {get_nested(stats, 'mbes_mesh', 'n_faces', default=0):,}")
    x_min = get_nested(stats, "mbes_mesh", "bounds", "x_min", default=0)
    x_max = get_nested(stats, "mbes_mesh", "bounds", "x_max", default=0)
    y_min = get_nested(stats, "mbes_mesh", "bounds", "y_min", default=0)
    y_max = get_nested(stats, "mbes_mesh", "bounds", "y_max", default=0)
    z_min = get_nested(stats, "mbes_mesh", "bounds", "z_min", default=0)
    z_max = get_nested(stats, "mbes_mesh", "bounds", "z_max", default=0)
    print(f"  Bounds (X): [{x_min:.2f}, {x_max:.2f}] m")
    print(f"  Bounds (Y): [{y_min:.2f}, {y_max:.2f}] m")
    print(f"  Bounds (Z): [{z_min:.2f}, {z_max:.2f}] m")

    # Camera
    print("\n📷 CAMERA CALIBRATION")
    focal = get_nested(stats, "camera_calibration", "focal_length", default=0)
    cx = get_nested(stats, "camera_calibration", "principal_point", default=0)
    width = get_nested(stats, "camera_calibration", "width_pixels", default=0)
    print(f"  Focal length: {focal:.2f} px")
    print(f"  Principal point: {cx:.2f} px")
    print(f"  Image width: {width:.0f} px")

    # Combined transect statistics (MAIN SECTION)
    if "aggregated_statistics" in stats:
        print("\n" + "=" * 80)
        print("📊 COMBINED TRANSECT STATISTICS (All H5 files together)")
        print("=" * 80)
        agg = stats["aggregated_statistics"]

        # File processing summary
        print(f"\n  Files processed: {get_nested(agg, 'total_files', default=0)} total")
        print(f"    ✅ Successful: {get_nested(agg, 'successful_files', default=0)}")
        print(f"    ❌ Failed: {get_nested(agg, 'failed_files', default=0)}")

        # HSI and ray tracing
        print(f"\n  HSI Data:")
        print(f"    Total frames: {get_nested(agg, 'total_frames', default=0):,}")
        print(f"    Total rays: {get_nested(agg, 'total_rays', default=0):,}")
        print(f"    Successful hits: {get_nested(agg, 'total_hits', default=0):,}")
        print(
            f"    Overall success rate: {get_nested(agg, 'overall_success_rate_pct', default=0):.2f}%"
        )

        # Mission metrics (combined)
        if "mission_metrics" in agg:
            print(f"\n  📏 Mission Metrics (entire transect):")
            dist_m = get_nested(agg, "mission_metrics", "total_distance_m", default=0)
            dur_sec = get_nested(
                agg, "mission_metrics", "total_duration_sec", default=0
            )
            speed_ms = get_nested(agg, "mission_metrics", "avg_speed_ms", default=0)
            speed_kmh = get_nested(agg, "mission_metrics", "avg_speed_kmh", default=0)
            print(f"    Total distance: {dist_m:.1f} m ({dist_m/1000:.3f} km)")
            print(f"    Total survey duration: {dur_sec:.1f} s ({dur_sec/60:.1f} min)")
            print(f"    Average speed: {speed_ms:.2f} m/s ({speed_kmh:.2f} km/h)")

        # Coverage metrics (combined)
        if "coverage_metrics" in agg:
            print(f"\n  📐 Coverage Metrics (entire transect):")
            area_m2 = get_nested(
                agg, "coverage_metrics", "total_coverage_area_m2", default=0
            )
            mean_swath = get_nested(
                agg, "coverage_metrics", "mean_swath_width_m", default=0
            )
            min_swath = get_nested(
                agg, "coverage_metrics", "min_swath_width_m", default=0
            )
            max_swath = get_nested(
                agg, "coverage_metrics", "max_swath_width_m", default=0
            )
            eff_pct = get_nested(
                agg, "coverage_metrics", "mean_coverage_efficiency_pct", default=0
            )
            print(
                f"    Total covered area: {area_m2:.1f} m² ({area_m2/10000:.4f} hectares)"
            )
            print(f"    Mean swath width: {mean_swath:.2f} m")
            print(f"    Swath width range: [{min_swath:.2f}, {max_swath:.2f}] m")
            print(f"    Coverage efficiency: {eff_pct:.1f}%")
            print(f"      (= covered area / survey bounding box area)")

    # Individual files
    if "individual_files" in stats and len(stats["individual_files"]) > 0:
        print("\n" + "=" * 80)
        print("📄 INDIVIDUAL FILE STATISTICS")
        print("=" * 80)

        for i, file_stat in enumerate(stats["individual_files"], 1):
            if file_stat.get("status") == "failed":
                print(
                    f"\n[{i}] {get_nested(file_stat, 'filename', default='Unknown')} ❌ FAILED"
                )
                continue

            print(f"\n[{i}] {get_nested(file_stat, 'filename', default='Unknown')}")
            print(
                f"  ⏱️  Processing time: {get_nested(file_stat, 'processing_time', default='N/A')}"
            )

            # HSI data
            if "hsi_data" in file_stat:
                hsi = file_stat["hsi_data"]
                n_frames = get_nested(hsi, "n_frames", default=0)
                n_slits = get_nested(hsi, "n_slits", default=0)
                n_bands = get_nested(hsi, "n_bands", default=0)
                duration = get_nested(hsi, "time_range", "duration_sec", default=0)
                time_start_file = get_nested(hsi, "time_range", "start", default=0)
                print(
                    f"  📸 HSI: {n_frames} frames × {n_slits} slits × {n_bands} bands"
                )
                print(f"      Duration: {duration:.1f} s")
                print(f"      Start: {unix_to_utc(time_start_file)}")

            # Navigation
            if "navigation" in file_stat:
                nav = file_stat["navigation"]
                print(f"  🧭 Navigation:")
                depth_min = get_nested(nav, "depth_range", "min", default=0)
                depth_max = get_nested(nav, "depth_range", "max", default=0)
                depth_mean = get_nested(nav, "depth_range", "mean", default=0)
                print(
                    f"      Depth: {depth_min:.2f} to {depth_max:.2f} m (mean: {depth_mean:.2f} m)"
                )

                roll_min = get_nested(nav, "attitude_ranges", "roll", "min", default=0)
                roll_max = get_nested(nav, "attitude_ranges", "roll", "max", default=0)
                roll_std = get_nested(nav, "attitude_ranges", "roll", "std", default=0)
                pitch_min = get_nested(
                    nav, "attitude_ranges", "pitch", "min", default=0
                )
                pitch_max = get_nested(
                    nav, "attitude_ranges", "pitch", "max", default=0
                )
                pitch_std = get_nested(
                    nav, "attitude_ranges", "pitch", "std", default=0
                )
                print(
                    f"      Roll: {roll_min:.2f}° to {roll_max:.2f}° (σ={roll_std:.2f}°)"
                )
                print(
                    f"      Pitch: {pitch_min:.2f}° to {pitch_max:.2f}° (σ={pitch_std:.2f}°)"
                )

            # Mission metrics
            if "mission_metrics" in file_stat:
                mission = file_stat["mission_metrics"]
                dist = get_nested(mission, "total_distance_m", default=0)
                speed_ms = get_nested(mission, "avg_speed_ms", default=0)
                speed_kmh = get_nested(mission, "avg_speed_kmh", default=0)
                print(f"  📏 Mission:")
                print(f"      Distance: {dist:.1f} m")
                print(f"      Speed: {speed_ms:.2f} m/s ({speed_kmh:.2f} km/h)")

            # Coverage
            if "coverage_metrics" in file_stat:
                coverage = file_stat["coverage_metrics"]
                area = get_nested(
                    coverage, "coverage_area", "total_covered_m2", default=0
                )
                swath_mean = get_nested(coverage, "swath_width", "mean_m", default=0)
                swath_std = get_nested(coverage, "swath_width", "std_m", default=0)
                eff = get_nested(coverage, "coverage_area", "efficiency_pct", default=0)
                print(f"  📐 Coverage:")
                print(f"      Area: {area:.1f} m²")
                print(f"      Swath width: {swath_mean:.2f} ± {swath_std:.2f} m")
                print(f"      Efficiency: {eff:.1f}%")

            # Ray tracing
            if "ray_tracing" in file_stat:
                rt = file_stat["ray_tracing"]
                success_pct = get_nested(rt, "success_rate_pct", default=0)
                hits = get_nested(rt, "successful_hits", default=0)
                total = get_nested(rt, "total_rays", default=1)
                trimesh_hits = get_nested(rt, "trimesh_hits", default=0)
                pv_retries = get_nested(rt, "pyvista_retries", default=0)
                pv_recoveries = get_nested(rt, "pyvista_recoveries", default=0)
                print(f"  🎯 Ray tracing:")
                print(f"      Success rate: {success_pct:.2f}% ({hits:,}/{total:,})")
                print(f"      Trimesh hits: {trimesh_hits:,}")
                print(
                    f"      PyVista retries: {pv_retries:,} (recovered: {pv_recoveries:,})"
                )

    print("\n" + "=" * 80)
    print(f"Statistics loaded from: {stats_file}")
    print("=" * 80)

    return stats


# Call the function
stats = print_processing_statistics()
```

## Comparison: Bounding Box vs Exact Footprint Masking

Let's compare both approaches side-by-side.

```python
# Side-by-side comparison
fig_comparison, axes = plt.subplots(1, 2, figsize=(16, 10), constrained_layout=True)

# LEFT: Bounding box approach (rectangular)
col_mask = (mb.x_cols >= x_bounds[0]) & (mb.x_cols <= x_bounds[1])
row_mask = (mb.y_rows >= y_bounds[0]) & (mb.y_rows <= y_bounds[1])
col_indices = np.where(col_mask)[0]
row_indices = np.where(row_mask)[0]
col_start, col_end = col_indices[0], col_indices[-1] + 1
row_start, row_end = row_indices[0], row_indices[-1] + 1
subregion_rect = mb.residuals[row_start:row_end, col_start:col_end]

sub_extent = (
    mb.x_cols[col_start] - mb.px_x / 2,
    mb.x_cols[col_end - 1] + mb.px_x / 2,
    mb.y_rows[row_end - 1] - mb.px_y / 2,
    mb.y_rows[row_start] + mb.px_y / 2,
)

vmin, vmax = mb._robust_sym_vlim(subregion_rect, q=0.98)

im1 = axes[0].imshow(
    np.ma.masked_invalid(subregion_rect),
    extent=sub_extent,
    origin="upper",
    cmap="RdBu_r",
    vmin=vmin,
    vmax=vmax,
)
axes[0].set_title("MBES - Bounding Box\n(Shows entire rectangular region)", fontsize=12)
axes[0].set_xlabel("Easting (m)")
axes[0].set_ylabel("Northing (m)")
fig_comparison.colorbar(im1, ax=axes[0], label="Residual depth (m)")

# RIGHT: Masked approach (exact UHI footprint)
# Create the mask again
coverage_mask = np.zeros(subregion_rect.shape, dtype=bool)
sub_y_rows = mb.y_rows[row_start:row_end]
sub_x_cols = mb.x_cols[col_start:col_end]

uhi_x_flat = uhi_x_coords.ravel()
uhi_y_flat = uhi_y_coords.ravel()
valid = np.isfinite(uhi_x_flat) & np.isfinite(uhi_y_flat)
uhi_x_flat = uhi_x_flat[valid]
uhi_y_flat = uhi_y_flat[valid]

pixel_buffer = max(mb.px_x, mb.px_y) / 2
for x_uhi, y_uhi in zip(uhi_x_flat, uhi_y_flat):
    col_idx = np.argmin(np.abs(sub_x_cols - x_uhi))
    row_idx = np.argmin(np.abs(sub_y_rows - y_uhi))
    x_dist = abs(sub_x_cols[col_idx] - x_uhi)
    y_dist = abs(sub_y_rows[row_idx] - y_uhi)
    if x_dist <= pixel_buffer and y_dist <= pixel_buffer:
        coverage_mask[row_idx, col_idx] = True

masked_subregion = np.ma.masked_where(~coverage_mask, subregion_rect)
masked_subregion = np.ma.masked_invalid(masked_subregion)

im2 = axes[1].imshow(
    masked_subregion,
    extent=sub_extent,
    origin="upper",
    cmap="RdBu_r",
    vmin=vmin,
    vmax=vmax,
)
axes[1].set_title(
    "MBES - Masked to UHI Coverage\n(Only shows where UHI measured)", fontsize=12
)
axes[1].set_xlabel("Easting (m)")
axes[1].set_ylabel("Northing (m)")
fig_comparison.colorbar(im2, ax=axes[1], label="Residual depth (m)")

plt.suptitle(
    f"MBES Comparison: Bounding Box vs Exact UHI Footprint Masking\nTracks {track_start}-{track_end}",
    fontsize=14,
    y=1.00,
)
plt.show()
```

## Investigate Coordinate System Alignment

Let's check if there's a rotation between the UHI and MBES coordinate systems.

```python
# Check MBES GeoTIFF coordinate reference system and orientation
import rasterio

with rasterio.open(config.MBES_GEOTIFF) as src:
    print("📊 MBES GeoTIFF Information:")
    print(f"   CRS: {src.crs}")
    print(f"   Transform:")
    print(f"      {src.transform}")
    print(f"   Bounds: {src.bounds}")
    print(f"   Shape: {src.shape} (rows, cols)")
    print(f"   Resolution: {src.res} (x, y)")

    # Check if north is up (standard) or if there's rotation
    affine = src.transform
    print(f"\n   Affine components:")
    print(f"      a (pixel width): {affine.a:.6f}")
    print(f"      b (row rotation): {affine.b:.6f}")
    print(f"      c (x origin): {affine.c:.2f}")
    print(f"      d (column rotation): {affine.d:.6f}")
    print(f"      e (pixel height): {affine.e:.6f}")
    print(f"      f (y origin): {affine.f:.2f}")

    if abs(affine.b) > 0.0001 or abs(affine.d) > 0.0001:
        print(
            f"\n   ⚠️  WARNING: MBES has rotation! (b={affine.b:.6f}, d={affine.d:.6f})"
        )
        rotation_deg = np.arctan2(affine.b, affine.a) * 180 / np.pi
        print(f"   Rotation angle: {rotation_deg:.4f}°")
    else:
        print(f"\n   ✅ MBES is axis-aligned (no rotation)")

# Check UHI coordinate alignment
print(f"\n📊 UHI Coordinate Information:")
print(f"   Track range: {track_start} to {track_end}")
print(f"   Coordinate shape: {uhi_x_coords.shape}")

# Calculate UHI principal axis (direction of travel)
# Use center line (middle slit) to see flight direction
mid_slit = uhi_x_coords.shape[1] // 2
x_centerline = uhi_x_coords[:, mid_slit]
y_centerline = uhi_y_coords[:, mid_slit]

# Remove NaNs
valid = np.isfinite(x_centerline) & np.isfinite(y_centerline)
x_centerline = x_centerline[valid]
y_centerline = y_centerline[valid]

if len(x_centerline) > 10:
    # Fit line to centerline to get flight direction
    coeffs = np.polyfit(x_centerline, y_centerline, 1)
    flight_angle_deg = np.arctan(coeffs[0]) * 180 / np.pi

    print(f"   Flight path centerline:")
    print(f"      Slope: {coeffs[0]:.6f}")
    print(f"      Angle from East: {flight_angle_deg:.4f}°")
    print(f"      (0° = East, 90° = North)")

    # Check if flight direction is aligned with UTM grid (North/South = 90°)
    deviation_from_north = abs(90 - abs(flight_angle_deg))
    if deviation_from_north > 5:
        print(
            f"   ⚠️  Flight path deviates {deviation_from_north:.2f}° from North/South"
        )
    else:
        print(f"   ✅ Flight path approximately North/South aligned")

    # Also check across-track direction
    # Take a slice across one track to see swath orientation
    mid_track = len(x_centerline) // 2
    x_swath = uhi_x_coords[mid_track, :]
    y_swath = uhi_y_coords[mid_track, :]
    valid_swath = np.isfinite(x_swath) & np.isfinite(y_swath)
    x_swath = x_swath[valid_swath]
    y_swath = y_swath[valid_swath]

    if len(x_swath) > 10:
        swath_coeffs = np.polyfit(x_swath, y_swath, 1)
        swath_angle_deg = np.arctan(swath_coeffs[0]) * 180 / np.pi
        print(f"\n   Swath (across-track) direction:")
        print(f"      Slope: {swath_coeffs[0]:.6f}")
        print(f"      Angle from East: {swath_angle_deg:.4f}°")
        print(f"      Expected: ~0° (East-West) for North-South flight")

        # Check perpendicularity
        angle_diff = abs(flight_angle_deg - swath_angle_deg)
        if 85 < angle_diff < 95:
            print(f"   ✅ Flight and swath are perpendicular ({angle_diff:.1f}°)")
        else:
            print(f"   ⚠️  Flight and swath not perpendicular ({angle_diff:.1f}°)")
```

### Quantify Rotation Angle

If there's a rotation, let's measure it quantitatively.

```python
# Quantify rotation between UHI and UTM grid
print("🔬 Rotation Analysis:\n")

# Method 1: Principal Component Analysis on UHI footprint
from sklearn.decomposition import PCA

# Get all valid UHI points
uhi_x_flat = uhi_x_coords.ravel()
uhi_y_flat = uhi_y_coords.ravel()
valid = np.isfinite(uhi_x_flat) & np.isfinite(uhi_y_flat)
uhi_points = np.column_stack([uhi_x_flat[valid], uhi_y_flat[valid]])

# Fit PCA to find principal axes
pca = PCA(n_components=2)
pca.fit(uhi_points)

# First principal component is the main direction (along-track)
pc1 = pca.components_[0]
pc2 = pca.components_[1]

# Angle of first principal component relative to East (positive X axis)
angle_pc1 = np.arctan2(pc1[1], pc1[0]) * 180 / np.pi
angle_pc2 = np.arctan2(pc2[1], pc2[0]) * 180 / np.pi

print(f"Method 1: Principal Component Analysis")
print(f"   PC1 (along-track) angle: {angle_pc1:.4f}° from East")
print(f"   PC2 (across-track) angle: {angle_pc2:.4f}° from East")
print(
    f"   Variance explained: PC1={pca.explained_variance_ratio_[0]*100:.1f}%, PC2={pca.explained_variance_ratio_[1]*100:.1f}%"
)

# Convert to compass bearing
bearing_pc1 = (90 - angle_pc1) % 360
print(f"   PC1 as compass bearing: {bearing_pc1:.2f}° (0°=North, 90°=East)")

# Method 2: Linear regression on centerline
mid_slit = uhi_x_coords.shape[1] // 2
x_center = uhi_x_coords[:, mid_slit]
y_center = uhi_y_coords[:, mid_slit]
valid_center = np.isfinite(x_center) & np.isfinite(y_center)
x_center = x_center[valid_center]
y_center = y_center[valid_center]

# Fit line: y = mx + b
if len(x_center) > 10:
    coeffs = np.polyfit(x_center, y_center, 1)
    slope = coeffs[0]
    angle_centerline = np.arctan(slope) * 180 / np.pi

    print(f"\nMethod 2: Centerline Linear Regression")
    print(f"   Slope: {slope:.6f}")
    print(f"   Angle from East: {angle_centerline:.4f}°")

    bearing_centerline = (90 - angle_centerline) % 360
    print(f"   As compass bearing: {bearing_centerline:.2f}°")

# Method 3: Check if MBES grid is rotated
with rasterio.open(config.MBES_GEOTIFF) as src:
    affine = src.transform
    mbes_rotation = np.arctan2(affine.b, affine.a) * 180 / np.pi

    print(f"\nMethod 3: MBES GeoTIFF Grid")
    print(f"   Rotation: {mbes_rotation:.6f}°")
    if abs(mbes_rotation) < 0.01:
        print(f"   → MBES grid is axis-aligned (North-South/East-West)")
    else:
        print(f"   → MBES grid has rotation!")

# Calculate deviation from North-South alignment
print(f"\n📊 Summary:")
deviation_from_north = abs(90 - abs(angle_pc1))
deviation_from_north_alt = abs(0 - abs(angle_pc1))
actual_deviation = min(deviation_from_north, deviation_from_north_alt)

if actual_deviation < 1.0:
    print(
        f"   ✅ UHI flight path is well-aligned with UTM grid ({actual_deviation:.2f}° deviation)"
    )
    print(f"   → No rotation correction needed")
elif actual_deviation < 5.0:
    print(f"   ⚠️  UHI has minor rotation: {actual_deviation:.2f}° from UTM grid")
    print(f"   → May be due to vehicle drift or navigation errors")
elif actual_deviation < 45.0:
    print(f"   ⚠️  UHI has significant rotation: {actual_deviation:.2f}° from UTM grid")
    print(f"   → Check if coordinate transformation is correct")
else:
    print(f"   ❌ UHI appears perpendicular to expected orientation!")
    print(f"   → Possible X/Y axis swap or coordinate system error")

# Additional check: Does the "tilt" you see match the measured angle?
print(f"\n💡 Visual Interpretation:")
print(f"   If you see a clockwise tilt in the UHI plot relative to MBES:")
print(f"   → UHI bearing should be < 180° (currently: {bearing_pc1:.2f}°)")
print(f"   If you see a counter-clockwise tilt:")
print(f"   → UHI bearing should be > 180° (currently: {bearing_pc1:.2f}°)")
print(f"   If perfectly aligned (no visible tilt):")
print(f"   → Bearing should be ~180° (North-South) or ~0°/360° (South-North)")
```
