## Plot with Navigation Trajectory

The `plot_georef_with_trajectory()` function extends the standard `plot_georef()` method by:
- ✅ Loading the full navigation trajectory from `config.NAV_CSV`
- ✅ Converting trajectory coordinates to the same coordinate system as the HSI plot
- ✅ Overlaying the trajectory line on top of the georeferenced HSI image
- ✅ Supporting all coordinate systems: LATLON, NED, and ECEF
- ✅ Interactive click-to-display coordinates (same as plot_georef)

**Customization options:**
- `trajectory_color`: Color of the trajectory line (e.g., "red", "cyan", "yellow")
- `trajectory_linewidth`: Width of the trajectory line
- `trajectory_alpha`: Transparency (0.0 to 1.0)
- `show_trajectory`: Set to `False` to hide trajectory
