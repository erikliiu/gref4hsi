````python
     normalize_method : str or None, optional
            Advanced normalization method for spectral offset removal:
            - None or False: No normalization (default, shows raw/corrected data)
            - "mean": Divide by mean (simple scaling, preserves amplitude differences)
            - "mean_center": Subtract mean (removes DC offset, preserves shape) ⭐ Recommended for offset removal
            - "snv": Standard Normal Variate (removes offset + scale, preserves shape) ⭐ Recommended for spectral comparison
            - "msc": Multiplicative Scatter Correction (physically meaningful, uses mean spectrum as reference)
            - "minmax": Min-max scaling to [0,1] (NOT recommended for spectral shape comparison)
            - "l2": L2 vector normalization (useful for clustering/PCA)
````

# TEST: Percentile Clipping

Test the new `apply_percentile_clipping()` function to remove extreme spikes before smoothing.

````python
# Helper function to visualize histograms per band
def test_plot_band_histograms(cube, sample_every_n_bands=10, figsize=(20, 12)):
    """
    Plot histograms for sampled wavelength bands to see distribution of values.
    This helps you identify which bands have extreme outliers (spikes).
    """
    n_bands = cube.data_corrected.shape[2]
    band_indices = list(range(0, n_bands, sample_every_n_bands))

    n_plots = len(band_indices)
    n_cols = 5
    n_rows = (n_plots + n_cols - 1) // n_cols

    fig, axes = plt.subplots(n_rows, n_cols, figsize=figsize)
    axes = axes.flatten()

    for idx, band_idx in enumerate(band_indices):
        if band_idx >= n_bands:
            break

        # Get all values for this band
        band_data = cube.data_corrected[:, :, band_idx].flatten()

        # Plot histogram
        axes[idx].hist(band_data, bins=50, alpha=0.7, edgecolor="black")
        axes[idx].set_title(
            f"Band {band_idx}\n{cube.wavelengths[band_idx]:.1f} nm", fontsize=10
        )
        axes[idx].set_xlabel("Intensity", fontsize=8)
        axes[idx].set_ylabel("Count", fontsize=8)
        axes[idx].grid(alpha=0.3)

        # Mark percentiles
        p5 = np.percentile(band_data, 5)
        p95 = np.percentile(band_data, 95)
        axes[idx].axvline(
            p5, color="red", linestyle="--", linewidth=1, alpha=0.7, label="5th %ile"
        )
        axes[idx].axvline(
            p95, color="red", linestyle="--", linewidth=1, alpha=0.7, label="95th %ile"
        )
        axes[idx].legend(fontsize=6)

    # Hide unused subplots
    for idx in range(len(band_indices), len(axes)):
        axes[idx].axis("off")

    plt.suptitle("Histogram per Band (sampled)", fontsize=14, fontweight="bold")
    plt.tight_layout()
    plt.show()


print("✅ test_plot_band_histograms() defined")
````

````python
# Step 1: Look at histograms BEFORE clipping
print("📊 Histograms BEFORE percentile clipping:")
test_plot_band_histograms(cube, sample_every_n_bands=20)
````

````python
# Step 2: Save one example spectrum BEFORE clipping (for comparison)
# Pick a pixel from the middle of the cube (safer than using ROI which might be out of bounds)
T, S, B = cube.data_corrected.shape
example_t = T // 2
example_s = S // 2
example_spectrum_before = cube.data_corrected[example_t, example_s, :].copy()
print(
    f"✅ Saved example spectrum from cube center at track={example_t}, slit={example_s}"
)
````

````python
# Step 4: Look at histograms AFTER clipping
print("📊 Histograms AFTER percentile clipping:")
test_plot_band_histograms(cube, sample_every_n_bands=20)
````

## Summary of Smoothing Methods

**Gaussian Smoothing** (`smoothing_method="gaussian"`):
- Good for general noise reduction
- `gaussian_sigma`: Larger = smoother (typical range: 1.0-3.0)
- Smooth, continuous results

**Median Smoothing** (`smoothing_method="median"`):
- Robust to outliers and spikes
- `median_kernel_size`: Larger = smoother (must be odd, typical: 3, 5, 7)
- Good for preserving edges while removing noise

**Whittaker Smoothing** (`smoothing_method="whittaker"`):
- Excellent feature preservation with penalized least squares
- `whittaker_lambda`: Larger = smoother (typical range: 1e3-1e5)
- Best for spectral contours and subtle features

**Bilateral Smoothing** (`smoothing_method="bilateral"`):
- Edge-preserving, intensity-aware smoothing
- `bilateral_sigma_spatial`: Controls spatial extent (typical: 2.0-3.0)
- `bilateral_sigma_intensity`: Controls edge preservation (smaller = stronger, typical: 0.05-0.2)
- Best for maintaining sharp transitions between features

**Recommendation**: Start with Whittaker for spectral analysis (excellent feature preservation), or Gaussian for quick noise reduction.
