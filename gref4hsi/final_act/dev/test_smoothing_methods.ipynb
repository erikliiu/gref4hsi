# Test: Smoothing Methods Comparison

This notebook tests the new smoothing options:
- `smoothing_method="moving_average"` (default, pandas rolling mean)
- `smoothing_method="savgol"` (Savitzky-Golay filter)

The Savitzky-Golay filter should preserve spectral features (peaks/valleys) better than simple moving average.

```python
import importlib
import sys
import os

sys.path.append(os.path.abspath("../"))

from utils.gref_pipeline import georef
from gref_pipeline import config

importlib.reload(georef)
from utils.gref_pipeline.georef import *
```

```python
# Load transect
transect = load_transect(config.OUTPUT_FOLDER)
transect.list_files()
```

```python
# Select file and apply illumination correction
cube = transect.select_files(["rad_uhi_20241029_115057_5"])
cube.describe()

cube.apply_illumination_correction_v2(window_size=500, force_recompute=False)
```

```python
# Import ROIs
cube.import_rois("my_rois_6.json")
cube.list_rois()
```

```python
# Define ROIs to plot
roi_names = [
    "sediment",
    "dark spots",
    "all bombs",
]
```

## Test 1: No Smoothing (baseline)

```python
cube.plot_spectrum(
    roi_names=roi_names,
    use_corrected=True,
    wavelength_range=(500, 650),
    wavelength_smoothing=1,  # No smoothing
    normalize_method=None,
    legend_loc="outside",
    use_inline_labels=False,
    show_std=True,
)
```

## Test 2: Moving Average (default)

```python
cube.plot_spectrum(
    roi_names=roi_names,
    use_corrected=True,
    wavelength_range=(500, 650),
    wavelength_smoothing=10,
    smoothing_method="moving_average",  # Default
    normalize_method=None,
    legend_loc="outside",
    use_inline_labels=False,
    show_std=True,
)
```

## Test 3: Savitzky-Golay (polyorder=2)

```python
cube.plot_spectrum(
    roi_names=roi_names,
    use_corrected=True,
    wavelength_range=(500, 650),
    wavelength_smoothing=10,
    smoothing_method="savgol",  # NEW: Savitzky-Golay
    savgol_polyorder=2,  # Quadratic polynomial
    normalize_method=None,
    legend_loc="outside",
    use_inline_labels=False,
    show_std=True,
)
```

## Test 4: Savitzky-Golay (polyorder=3)

```python
cube.plot_spectrum(
    roi_names=roi_names,
    use_corrected=True,
    wavelength_range=(500, 650),
    wavelength_smoothing=10,
    smoothing_method="savgol",  # NEW: Savitzky-Golay
    savgol_polyorder=3,  # Cubic polynomial
    normalize_method=None,
    legend_loc="outside",
    use_inline_labels=False,
    show_std=True,
)
```

## Test 5: Comparison with larger window (20)

```python
# Moving Average with window=20
cube.plot_spectrum(
    roi_names=roi_names,
    use_corrected=True,
    wavelength_range=(500, 650),
    wavelength_smoothing=20,
    smoothing_method="moving_average",
    normalize_method=None,
    legend_loc="outside",
    use_inline_labels=False,
    show_std=False,
)
```

```python
# Savitzky-Golay with window=20
cube.plot_spectrum(
    roi_names=roi_names,
    use_corrected=True,
    wavelength_range=(500, 650),
    wavelength_smoothing=20,
    smoothing_method="savgol",
    savgol_polyorder=2,
    normalize_method=None,
    legend_loc="outside",
    use_inline_labels=False,
    show_std=False,
)
```
