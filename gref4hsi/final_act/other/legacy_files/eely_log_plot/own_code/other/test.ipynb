```python
import numpy as np
import matplotlib.pyplot as plt

# Time values
t = np.linspace(0, 1, 5, endpoint=False)

# Original signal: A sine wave
signal = np.sin(2 * np.pi * t)

# Plot
plt.plot(t, signal, "o-", label="Original Signal")
plt.legend()
plt.show()
```

```python
fft_signal = np.fft.fft(signal)
```

```python
import numpy as np

# Create a 3D NumPy array with shape (5, 10, 20)
data = np.random.rand(5, 10, 20)

# Get the shape
print(data.shape)  # Output: (5, 10, 20)

# Unpacking the shape
n_tracks, n_slit_pixels, n_wavelengths = data.shape
print(n_tracks)  # 5
print(n_slit_pixels)  # 10
print(n_wavelengths)  # 20
```
