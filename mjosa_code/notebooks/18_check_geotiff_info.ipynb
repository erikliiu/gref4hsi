```python
def check_geotiff_info(geotiff_path):
    """
    Extract and display GeoTIFF metadata including grid cell size.
    """
    if not geotiff_path.exists():
        print(f"❌ File not found: {geotiff_path}")
        return

    print(f"\n{'='*80}")
    print(f"📊 GeoTIFF Info: {geotiff_path.name}")
    print(f"{'='*80}")

    with rasterio.open(geotiff_path) as src:
        # Basic info
        print(f"\n📐 Dimensions:")
        print(f"   Width (columns):  {src.width} pixels")
        print(f"   Height (rows):    {src.height} pixels")
        print(f"   Bands:            {src.count}")
        print(f"   Data type:        {src.dtypes[0]}")

        # Coordinate system
        print(f"\n🌍 Coordinate System:")
        if src.crs:
            epsg = src.crs.to_epsg()
            print(f"   EPSG:             {epsg}")
            print(f"   CRS:              {src.crs}")
        else:
            print(f"   ⚠️  No CRS defined")

        # Grid cell size (RESOLUTION)
        print(f"\n📏 Grid Cell Size (Resolution):")
        transform = src.transform
        cell_size_x = abs(transform.a)  # Pixel width in CRS units
        cell_size_y = abs(transform.e)  # Pixel height in CRS units
        print(f"   X (longitude/easting):  {cell_size_x:.6f} meters")
        print(f"   Y (latitude/northing):  {cell_size_y:.6f} meters")

        if abs(cell_size_x - cell_size_y) < 0.0001:
            print(f"   ✅ Square pixels: {cell_size_x:.6f} m")
        else:
            print(f"   ⚠️  Non-square pixels")

        # Bounds
        print(f"\n🗺️  Bounding Box (in CRS units):")
        print(f"   Left (min X):     {src.bounds.left:.2f}")
        print(f"   Right (max X):    {src.bounds.right:.2f}")
        print(f"   Bottom (min Y):   {src.bounds.bottom:.2f}")
        print(f"   Top (max Y):      {src.bounds.top:.2f}")
        print(f"   Width:            {src.bounds.right - src.bounds.left:.2f} meters")
        print(f"   Height:           {src.bounds.top - src.bounds.bottom:.2f} meters")

        # Transform details
        print(f"\n🔧 Affine Transform:")
        print(f"   {transform}")
        print(f"\n   Transform breakdown:")
        print(f"   - a (pixel width):     {transform.a:.6f}")
        print(f"   - b (row rotation):    {transform.b:.6f}")
        print(f"   - c (x offset):        {transform.c:.2f}")
        print(f"   - d (column rotation): {transform.d:.6f}")
        print(f"   - e (pixel height):    {transform.e:.6f}")
        print(f"   - f (y offset):        {transform.f:.2f}")

        # Nodata value
        print(f"\n🚫 NoData Value:")
        print(f"   {src.nodata}")

    print(f"\n{'='*80}\n")
```

```python
# Check full MBES coverage
check_geotiff_info(config.MBES_GEOTIFF)
```
